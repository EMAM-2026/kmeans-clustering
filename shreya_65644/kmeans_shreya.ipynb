```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
df=pd.read_csv('Mall_Customers.csv')
print(df.head())
print(df.dtypes)
print(df.shape)
print(df.isnull().sum())
```

```python
X=df[['Annual Income (k$)','Spending Score (1-100)']]
plt.scatter(X.iloc[:,0],X.iloc[:,1])
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.show()
```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
print("Original feature means:")
display(X.mean().to_frame("mean"))
print("Original feature standard deviations:")
display(X.std(ddof=0).to_frame("std_dev"))
print("Standardized means:",X_scaled.mean(axis=0).round(3))
print("Standardized standard deviations:",X_scaled.std(axis=0).round(3))
```

```python
wcss=[]
for i in range(1,11):
    kmeans=KMeans(n_clusters=i,init='k-means++',random_state=42,n_init=10)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

plt.plot(range(1,11),wcss,marker='o')
plt.xlabel('k')
plt.ylabel('WCSS')
plt.show()
```

```python
kmeans=KMeans(n_clusters=5,init='k-means++',random_state=42,n_init=10)
df['Cluster']=kmeans.fit_predict(X_scaled)
plt.scatter(X.iloc[:,0],X.iloc[:,1],c=df['Cluster'])
centroids=scaler.inverse_transform(kmeans.cluster_centers_)
plt.scatter(centroids[:,0],centroids[:,1],s=75,c='red',marker='x')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.show()
```

```python
summary=df.groupby('Cluster').agg({
    'CustomerID':'count',
    'Annual Income (k$)':'mean',
    'Spending Score (1-100)':'mean'
}).rename(columns={'CustomerID':'Count'})
print(summary)
```
